## 2. Data Collection Methodology

### Challenges Encountered
- The official website (https://www.livingstonecouncil.gov.zm) has an **invalid SSL certificate** (`ERR_CERT_DATE_INVALID`). Standard HTTPS requests fail.
- Consequently, primary data was collected via:
  1. Direct inspection of publicly cached / indexed content from the Civic Leaders page.
  2. Official Livingstone City Council Facebook posts and press statements.
  3. Reputable secondary sources (Diggers News, ZNBC, UNICEF budget briefs, MLGRD listings).

### Ethical & Academic Notes
- Only publicly available information was used.
- Every record includes a `source` and `source_url` column for transparency and reproducibility.
- Amounts were standardised to numeric ZMW values where possible.

## 3. Civic Leaders Dataset

Data sourced from the official Civic Leaders page of Livingstone City Council.

```python
# Load the curated civic leaders data
civic = pd.read_csv(
    'db-unza26-csc4792-livingstone_civic_leaders.csv',
    sep='|'
)

print("Shape:", civic.shape)
print("\nColumn names:", list(civic.columns))
civic.head(10)
```

```python
# Basic exploration
print("Number of records:", len(civic))
print("\nPosition types:")
print(civic['position_type'].value_counts())
print("\nUnique wards represented:")
print(civic['ward'].dropna().unique())
```

## 4. CDF Projects Dataset

Compiled from official council announcements, Facebook posts, and verified news reports (2022–2026).

```python
cdf = pd.read_csv(
    'db-unza26-csc4792-livingstone_cdf_projects.csv',
    sep='|'
)

print("Shape:", cdf.shape)
cdf.head()
```

```python
# Convert amount to numeric where possible
cdf['amount_numeric'] = pd.to_numeric(cdf['amount_zmw'], errors='coerce')

print("Projects with quantified amounts:")
print(cdf[['project_name', 'year', 'amount_numeric', 'category', 'status']].dropna(subset=['amount_numeric']))

print("\nTotal quantified CDF spend captured (ZMW):", cdf['amount_numeric'].sum())
print("\nProjects by category:")
print(cdf['category'].value_counts())
```

## 7. Data Cleaning & Preprocessing Steps Performed

1. **Standardisation of amounts**: Removed currency symbols and commas; converted to pure numeric where possible.
2. **Ward name normalisation**: Corrected spelling variations (e.g., "Mosi ao tunya" → "Mosi-oa-Tunya").
3. **Missing value handling**: Explicitly left blank or marked "Not specified" rather than imputing.
4. **Source attribution**: Every row carries `source` and `source_url` for auditability.
5. **Date stamping**: `scraped_date` column added to all files.
6. **File format compliance**: All output CSVs use `|` as separator and follow the required naming convention `db-unza26-csc4792-[DESCRIPTION].csv`.

## 9. Limitations

- Official website SSL certificate issues prevented fully automated scraping of all pages.
- Not all CDF projects have publicly disclosed exact monetary values.
- Some ward-level project details are incomplete in public sources.
- Delimitation of Livingstone into East/West constituencies was under discussion in 2026; ward lists may evolve.
- Financial data (full approved budgets, LGEF utilisation, local revenue) is not comprehensively published on the council website.

These limitations are documented so that future researchers understand the coverage of the dataset.
